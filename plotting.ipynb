```python
##########################################################################################
# Dependencies
import math
import numpy as np
from sklearn.covariance import EmpiricalCovariance
from scipy.stats import ortho_group
from tqdm import tqdm
import matplotlib.pyplot as plt
import os
from scipy.stats import norm

from typing import Callable



plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 30

# load the pkl file
import pickle
with open('./results/dists_gradient.pkl', 'rb') as f:
    results = pickle.load(f)

fig, ax = plt.subplots(figsize=(10, 8))
for key, dist in results.items():
    H, n_values = key
    ax.plot(range(len(dist)), np.array(dist), label=f'$H={H}$, $n_{{\\mathrm{{values}}}}={n_values}$')
ax.set_ylabel("Distance Estimate")
ax.set_xlabel("Time")
# # plt.legend(fontsize=14)
ax.legend(fontsize=20, loc='upper right')
plt.tight_layout()
plt.savefig("./distr_steer_3.pdf")

```
